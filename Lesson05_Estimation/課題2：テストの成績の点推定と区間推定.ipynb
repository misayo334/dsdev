```python
import pandas as pd
import numpy as np

from scipy import stats

import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme()

#csvファイル読み込み
df = pd.read_csv("./students_scores.csv")
#中身を表示してみる
df
```

```python
#english列のみ抽出してndarray型にする
english_score = np.array(df["english"])
#データ型を表示
print(english_score.dtype)
#中身を表示してみる
english_score
```

```python
# matplotlibでヒストグラムを作成
plt.hist(english_score, bins=20)
```

```python
# seabornでヒストグラムと分布を作成
sns.histplot(english_score, kde=True, bins=20)
```

```python
#母平均
eng_mean = np.mean(english_score)
print("母平均: ",eng_mean)

#母分散
eng_var = np.var(english_score, ddof=1)  #不偏分散を使う
print("母分散: ",eng_var)

#標準誤差
eng_std = np.std(english_score, ddof=1)  #不偏標準偏差
eng_se = eng_std/np.sqrt(len(english_score))
print("標準誤差: ",eng_se)

#95%信頼区間の下限値と上限値
# 自由度（標本数 - 1）を計算
eng_df = len(english_score) - 1  #( n-1) = degrees of freedom　自由度
# 95%信頼区間の下限値と上限値を計算
interval = stats.t.interval(0.95, eng_df, loc=eng_mean, scale=eng_se)
print(interval)
```
